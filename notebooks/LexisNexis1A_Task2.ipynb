```python
df["Make"] = df["Make"].replace({"Accura": "Acura", "Nisson": "Nissan", "Porche": "Porsche", "Mecedes": "Mercedes"})
df["Make"].unique()
```

```python
print(df[df["AgeOfPolicyHolder"] == "16 to 17"]["Age"].value_counts())
print(df.loc[1516, ["Age", "FraudFound_P"]])
```

```python
# drop the one bad row (row 1516, not a fraud case)
df = df.drop(index=1516)

# treat Age = 0 as missing and flag it, instead of filling in an age
df["Age_missing"] = (df["Age"] == 0).astype(int)
df["Age"] = df["Age"].mask(df["Age"] == 0)

# check that it worked
print(df.shape)
print(df["Age_missing"].sum())
print(df["Age"].isna().sum())
print(df["Age"].min())
print(df["DayOfWeekClaimed"].nunique(), df["MonthClaimed"].nunique())
print(df["FraudFound_P"].value_counts())
```

```python
print(df.groupby("AgeOfPolicyHolder")["Age"].agg(["min", "max", "count"]))
```

```python
bins = [0, 17, 20, 25, 30, 35, 40, 50, 65, 120]
labels = ["16 to 17", "18 to 20", "21 to 25", "26 to 30", "31 to 35", "36 to 40", "41 to 50", "51 to 65", "over 65"]
known = df[df["Age"].notna()]
expected = pd.cut(known["Age"], bins=bins, labels=labels)
print((expected.astype(str) != known["AgeOfPolicyHolder"]).sum(), len(known))
```

```python
def fraud_rate_table(data, col):
    t = data.groupby(col)["FraudFound_P"].agg(claims="count", fraud="sum", fraud_rate="mean")
    t["fraud_rate"] = (t["fraud_rate"] * 100).round(1)
    return t.sort_values("fraud_rate", ascending=False)

for col in ["Age_missing", "PoliceReportFiled", "WitnessPresent", "NumberOfSuppliments",
            "AddressChange_Claim", "PastNumberOfClaims", "Days_Policy_Accident", "Days_Policy_Claim"]:
    print(fraud_rate_table(df, col), "\n")

print(df.groupby("Year")["PolicyNumber"].agg(["min", "max", "count"]))
```
